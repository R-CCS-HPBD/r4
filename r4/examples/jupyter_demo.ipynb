# R4 Jupyter Integration Demo

This notebook demonstrates the Jupyter integration features of the `r4` SDK for the R4 Platform.

## Features Covered

- **Magic Commands**: `%%r4`, `%rag`, `%%rag_chat`, `%r4_config`
- **Widgets**: `ChatWidget`, `RAGWidget`
- **Output Formatting**: Rich HTML/Markdown rendering

```python
# Set the RAG collection
%r4_config collection=my-research-papers
```

## 3. Magic Commands

### 3.1 Chat with LLM (`%%r4`)

Use `%%r4` to send prompts to the LLM:

```python
%%r4
What is the difference between supervised and unsupervised learning?
Please provide a brief explanation with examples.
```

```python
%%r4 --no-stream
Explain the concept of backpropagation in neural networks.
```

### 3.2 RAG Search (`%rag`)

Search your document collections with `%rag`:

```python
# Search with default settings
%rag quantum entanglement simulation
```

```python
# Search with custom parameters
%rag --collection physics-papers --top-k 10 molecular dynamics force field
```

### 3.3 RAG-Augmented Chat (`%%rag_chat`)

Combine RAG retrieval with LLM generation:

```python
%%rag_chat
Based on my research papers, what are the main findings about
protein folding simulations? Summarize the key results.
```

```python
%%rag_chat --collection=simulation-results --top-k=5
What computational methods were used in the climate modeling study?
```

## 4. Interactive Widgets

### 4.1 Chat Widget

Create an interactive chat interface:

```python
from r4 import get_chat_widget

# Create a chat widget with a system prompt
chat = get_chat_widget(
    model="rivault-gpt-oss-static",
    system_prompt="You are a helpful scientific assistant specializing in computational physics.",
)

# Display widget info
chat.display()
```

```python
# Send a message
response = chat.chat("What are the main challenges in molecular dynamics simulations?")
```

```python
# Continue the conversation
response = chat.chat("How does parallel computing help address these challenges?")
```

```python
# View chat history
chat.history()
```

```python
# Clear history to start fresh
chat.clear()
```

### 4.2 RAG Widget

Create an interactive RAG interface:

```python
from r4 import get_rag_widget

# Create a RAG widget
rag = get_rag_widget(collection="research-papers", top_k=5)

# Display widget info
rag.display()
```

```python
# Search documents
results = rag.query("neural network optimization techniques")
```

```python
# Ask a question with RAG augmentation
answer = rag.ask("What optimization algorithms are recommended for deep learning?")
```

## 5. Output Formatting

Use the `OutputFormatter` for rich display:

```python
from r4 import get_output_formatter

fmt = get_output_formatter()
```

```python
# Display Markdown
fmt.markdown("""
# Results Summary

The simulation completed successfully with the following metrics:

- **Accuracy**: 95.2%
- **F1 Score**: 0.94
- **Training Time**: 2.5 hours
""")
```

```python
# Display code with syntax highlighting
fmt.code(
    """
import numpy as np
from scipy.integrate import odeint

def simulate_oscillator(y, t, omega):
    x, v = y
    dydt = [v, -omega**2 * x]
    return dydt

# Run simulation
t = np.linspace(0, 10, 1000)
y0 = [1.0, 0.0]
solution = odeint(simulate_oscillator, y0, t, args=(2*np.pi,))
""",
    language="python",
)
```

```python
# Display chat-style messages
fmt.chat_message("user", "Can you analyze these simulation results?")
fmt.chat_message(
    "assistant",
    "I'd be happy to analyze the results. The data shows a clear periodic pattern with decreasing amplitude, consistent with damped harmonic oscillation.",
)
```

```python
# Display status messages
fmt.info("Processing data...")
fmt.success("Analysis complete!")
fmt.error("Failed to connect to remote server")
```

## 6. Advanced Usage

### 6.2 Store Results in Variables

```python
# The last response is stored in the _ variable
# You can also capture it explicitly

chat = get_chat_widget()
analysis = chat.chat(
    "Summarize the key trends in AI for scientific computing in 2025.", stream=False
)

# Now 'analysis' contains the response text
print(f"Response length: {len(analysis)} characters")
```
